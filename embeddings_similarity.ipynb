```python
import numpy as np
import pandas as pd
import torch
from sentence_transformers import SentenceTransformer, util

print("Libraries imported successfully!")
print("PyTorch version:", torch.__version__)
```

```python
model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    cache_folder="./model_cache"
)

print("Embedding model loaded successfully!")
```

```python
texts = [
    "I love learning artificial intelligence.",
    "I enjoy studying AI.",
    "Python is a programming language.",
    "I want to build machine learning models.",
    "My bicycle has a flat tyre."
]

for i, text in enumerate(texts, start=1):
    print(f"Sentence {i}: {text}")
```

```python
embeddings = model.encode(texts)

print("Number of sentences:", len(texts))
print("Embedding shape:", embeddings.shape)
print("\nFirst five values of the first embedding:")
print(embeddings[0][:5])
```

```python
similarity = util.cos_sim(
    embeddings[0],
    embeddings[1]
)

print("Sentence 1:", texts[0])
print("Sentence 2:", texts[1])
print("Cosine similarity:", round(similarity.item(), 3))
```

```python
similarity_matrix = util.cos_sim(
    embeddings,
    embeddings
)

similarity_df = pd.DataFrame(
    similarity_matrix.numpy(),
    index=[f"Sentence {i}" for i in range(1, len(texts) + 1)],
    columns=[f"Sentence {i}" for i in range(1, len(texts) + 1)]
)

display(similarity_df.round(3))
```

```python
# Convert the similarity matrix into a NumPy array.
scores = similarity_matrix.numpy()

# Ignore self-comparisons by considering only the upper triangle.
upper_triangle = np.triu(scores, k=1)

# Find the indices of the largest remaining similarity score.
i, j = np.unravel_index(
    np.argmax(upper_triangle),
    upper_triangle.shape
)

print("Most similar sentence pair:")
print(f"Sentence {i + 1}: {texts[i]}")
print(f"Sentence {j + 1}: {texts[j]}")
print("Similarity score:", round(float(scores[i, j]), 3))
```

```python
print("SIMILARITY ANALYSIS\n")

for i in range(len(texts)):
    for j in range(i + 1, len(texts)):
        score = float(scores[i, j])

        print(f"Sentence {i + 1} vs Sentence {j + 1}: {score:.3f}")

print("\nConclusion:")
print("Higher cosine similarity indicates more similar embedding directions.")
print("Similarity scores estimate semantic relatedness, not exact meaning or accuracy.")
```
